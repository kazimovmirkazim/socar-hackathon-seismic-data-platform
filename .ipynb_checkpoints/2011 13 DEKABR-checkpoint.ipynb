```python
import os
import struct
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

# --- Qovluq ---
folder = r"./track_1_forensics/"

# --- SGX fayllarını oxu və Parquet-ə çevir ---
sgx_files = [f for f in os.listdir(folder) if f.endswith(".sgx")]
for sgx in sgx_files:
    path = os.path.join(folder, sgx)
    traces = []
    
    with open(path, "rb") as f:
        header = f.read(16)
        magic, survey_id, trace_count = struct.unpack("<8sII", header)
        
        for _ in range(trace_count):
            trace_bytes = f.read(13)
            if len(trace_bytes) < 13:
                break  # corrupted end
            well_id, depth, amplitude, quality = struct.unpack("<IffB", trace_bytes)
            traces.append({
                "well_id": well_id,
                "depth": depth,
                "amplitude": amplitude,
                "quality": quality,
                "survey_id": survey_id
            })

    if traces:
        df = pd.DataFrame(traces)
        parquet_out = os.path.join(folder, sgx.replace(".sgx", ".parquet"))
        table = pa.Table.from_pandas(df)
        pq.write_table(table, parquet_out)
        print(f"{sgx} -> {parquet_out} çevrildi. Rows: {len(df)}")
```

```python
import os
import struct
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

folder = r"./track_1_forensics"
sgx_files = [f for f in os.listdir(folder) if f.endswith(".sgx")]

for sgx in sgx_files:
    path = os.path.join(folder, sgx)
    traces = []

    with open(path, "rb") as f:
        # Try to read header
        try:
            header = f.read(16)
            magic, survey_id, trace_count = struct.unpack("<8sII", header)
        except:
            print(f"{sgx} header oxunmadı, keçilir.")
            continue

        # Read traces until EOF
        while True:
            trace_bytes = f.read(13)
            if len(trace_bytes) < 13:
                break  # faylın sonuna çatdıq
            try:
                well_id, depth, amplitude, quality = struct.unpack("<IffB", trace_bytes)
                traces.append({
                    "well_id": well_id,
                    "depth": depth,
                    "amplitude": amplitude,
                    "quality": quality,
                    "survey_id": survey_id
                })
            except:
                continue  # corrupted trace keçilir

    # Parquet-ə yaz
    if traces:
        df = pd.DataFrame(traces)
        parquet_out = os.path.join(folder, sgx.replace(".sgx", ".parquet"))
        try:
            table = pa.Table.from_pandas(df)
            pq.write_table(table, parquet_out)
            print(f"{sgx} -> {parquet_out} çevrildi. Rows: {len(df)}")
        except Exception as e:
            print(f"{sgx} Parquet yazılmadı:", e)
```

```python
def simple_footer_fix(file_path, output_path=None):
    """Sadəcə son 4 byte-ı PAR1 ilə əvəz edir"""
    
    if output_path is None:
        output_path = file_path.replace('.parquet', '_fixed.parquet')
    
    with open(file_path, 'rb') as f:
        data = f.read()
    
    print(f"Original size: {len(data)} bytes")
    print(f"Original last 8 bytes: {data[-8:].hex()} = {repr(data[-8:])}")
    
    # Sadəcə son 4 byte-ı PAR1 ilə əvəz et
    new_data = data[:-4] + b'PAR1'
    
    with open(output_path, 'wb') as f:
        f.write(new_data)
    
    print(f"\n✓ Created: {output_path}")
    print(f"New last 8 bytes: {new_data[-8:].hex()} = {repr(new_data[-8:])}")
    
    # Test et
    try:
        import pandas as pd
        df = pd.read_parquet(output_path)
        print(f"\n✓✓✓ SUCCESS! ✓✓✓")
        print(f"Rows: {len(df)}")
        print(f"Columns: {list(df.columns)}")
        print(f"\nFirst 5 rows:")
        print(df.head())
        return output_path
    except Exception as e:
        print(f"\n Failed: {str(e)[:200]}")
        return None

# ƏN SADƏ HƏLL - BU İŞLƏYƏ BİLƏR
fixed_file = simple_footer_fix(file_path)
```

```python
import os
import struct
import pandas as pd

def repair_parquet_footer_minimal(file_path, output_path=None):
    """
    Parquet faylının pozulmuş footer-ini təmir edir.
    Standart struktur: [PAR1][data][metadata][metadata_size(4bytes)][PAR1]
    Təmir: Əgər son 4 bayt pozulubsa, onu PAR1 ilə əvəz edir.
    """
    
    if output_path is None:
        # Təmir olunmuş faylın yolunu təyin edir
        output_path = file_path.replace('.parquet', '_repaired.parquet')
    
    with open(file_path, 'rb') as f:
        data = f.read()
    
    file_size = len(data)
    
    # Parquet Header-i yoxla
    if data[:4] != b'PAR1':
        # Header pozulubsa, təmiri mümkün deyil
        return None
    
    # Standart olaraq, faylın son 4 baytı PAR1 olmalıdır.
    # Əgər pozulma son 4 baytda baş veribsə:
    if data[-4:] != b'PAR1':
        # Faylın son 4 baytını kəs (pozulmuş hissəni at) və PAR1 əlavə et
        new_data = data[:-4] + b'PAR1'
        
        try:
            with open(output_path, 'wb') as f:
                f.write(new_data)
        except IOError:
            return None
        
        # Təmir olunmuş faylı test et
        try:
            df = pd.read_parquet(output_path)
            # Əgər fayl uğurla oxunubsa, bərpa edilmiş yolu qaytar
            return output_path
        except Exception:
            # Hələ də oxunmursa, təmiri uğursuz say
            return None
    
    # Faylın footer-i artıq düzgündür
    return file_path

# İSTİFADƏ NÜMUNƏSİ (Hackathon Skriptində bu hissə dinamik olmalıdır)
# # Əgər fayl yolu verilsə, aşağıdakı kimi işlədilə bilər:
# # file_to_repair = "/path/to/archive_batch_seismic_readings_recovered.parquet"
# # repaired_file = repair_parquet_footer_minimal(file_to_repair)
# # if repaired_file:
# #     print(f"File repaired and saved as: {repaired_file}")
```

```python
import os
import struct

def repair_parquet_footer(file_path, output_path=None):
    """Parquet faylının pozulmuş footer-ini təmir edir"""
    
    if output_path is None:
        output_path = file_path.replace('.parquet', '_repaired.parquet')
    
    print(f"Reading file: {file_path}")
    
    with open(file_path, 'rb') as f:
        data = f.read()
    
    file_size = len(data)
    print(f"File size: {file_size:,} bytes")
    
    # Header yoxla
    if data[:4] != b'PAR1':
        print(" Header also corrupted!")
        return None
    
    print("✓ Header OK (PAR1)")
    
    # Footer-i tap - PAR1 olmalı son 4 byte-da
    # Parquet structure: [PAR1][data][metadata][metadata_size(4bytes)][PAR1]
    
    # Geriyə doğru PAR1 axtar
    print("\nSearching for PAR1 markers in file...")
    par1_positions = []
    for i in range(len(data) - 3):
        if data[i:i+4] == b'PAR1':
            par1_positions.append(i)
    
    print(f"Found {len(par1_positions)} PAR1 markers at positions: {par1_positions}")
    
    if len(par1_positions) < 2:
        print(" Not enough PAR1 markers to reconstruct")
        return None
    
    # İlk PAR1 header (position 0)
    # İkinci PAR1 footer olmalıdır
    
    # Metadata size-ı tap (footer-dən əvvəlki 4 byte)
    # Son PAR1-dən əvvəl gələn 4 byte metadata size-dır
    
    # Strategiya: Son 100 byte-da metadata+size+PAR1 axtaraq
    print("\n--- Analyzing file end ---")
    print(f"Last 50 bytes (hex): {data[-50:].hex()}")
    print(f"Last 50 bytes (repr): {repr(data[-50:])}")
    
    # Yeni footer qur
    # Metadata başlanğıcını tap
    # Parquet faylında metadata-dan əvvəl data bloků var
    
    # Daha asan yol: Metadata-nı tap və düzgün footer əlavə et
    
    # Faylın strukturunu analiz et
    print("\n--- Attempting reconstruction ---")
    
    # Strategiya: Faylı PAR1-dən sonrakı ikinci PAR1-ə qədər götür
    # və sonra düzgün footer əlavə et
    
    # Sadə yol: Əgər footer sadəcə pozulubsa, onu düzəlt
    
    # Son 8 byte nədir?
    print(f"\nLast 8 bytes: {data[-8:].hex()} = {repr(data[-8:])}")
    
    # Metadata size-ı hesabla (footer-dən 8 byte əvvəl)
    try:
        # Metadata size little-endian 4-byte integer
        metadata_size_bytes = data[-8:-4]
        metadata_size = struct.unpack('<I', metadata_size_bytes)[0]
        print(f"Metadata size (from -8:-4): {metadata_size}")
        
        # Metadata başlangıcı
        metadata_start = file_size - 8 - metadata_size
        print(f"Metadata should start at: {metadata_start}")
        
        if metadata_start > 0 and metadata_start < file_size:
            # Yeni fayl qur: [header data metadata][metadata_size][PAR1]
            new_data = data[:file_size-4] + b'PAR1'
            
            with open(output_path, 'wb') as f:
                f.write(new_data)
            
            print(f"\n✓ Created repaired file: {output_path}")
            
            # Test et
            import pandas as pd
            try:
                df = pd.read_parquet(output_path)
                print(f"✓ SUCCESS! Read {len(df)} rows, {len(df.columns)} columns")
                print(f"Columns: {list(df.columns)}")
                return output_path
            except Exception as e:
                print(f" Still cannot read: {e}")
                
    except Exception as e:
        print(f"Error in reconstruction: {e}")
    
    # Alternativ: Faylı blok-blok oxuyub yenidən yaz
    print("\n--- Trying block reconstruction ---")
    try_block_reconstruction(file_path, output_path)
    
    return None

def try_block_reconstruction(file_path, output_path):
    """Faylı blok-blok oxuyub yenidən qurmağa çalışır"""
    
    import pyarrow.parquet as pq
    
    with open(file_path, 'rb') as f:
        data = f.read()
    
    # Müxtəlif kəsmə nöqtələri sına
    for cut_point in range(len(data) - 100, len(data) - 4):
        try:
            # Son hissəni kəs və PAR1 əlavə et
            test_data = data[:cut_point]
            
            # Metadata size-ı yenidən hesabla
            # Metadata footer-dən 8 byte əvvəl başlayır deyə
            potential_metadata_size = cut_point - 4  # təxmini
            
            # Düzgün struktur qur
            # Amma bu çox çətindir, daha sadə yol...
            
            pass  # Bu metod çox mürəkkəbdir
            
        except:
            pass
    
    print("Block reconstruction failed")

# İŞLƏT
file_path = r"./data_lake/raw/archive_batch_seismic_readings.parquet"
repaired = repair_parquet_footer(file_path)
```

```python
def extract_valid_parquet_and_flag(file_path):
    """Parquet-i və flag-ı ayırır"""
    
    with open(file_path, 'rb') as f:
        data = f.read()
    
    print(f"Total file size: {len(data)} bytes")
    
    # PAR1 markers
    par1_positions = []
    for i in range(len(data) - 3):
        if data[i:i+4] == b'PAR1':
            par1_positions.append(i)
    
    print(f"PAR1 positions: {par1_positions}")
    
    # İkinci PAR1 (position 40040) - bu footer marker olmalıdır
    second_par1 = par1_positions[1]
    print(f"\nSecond PAR1 at position: {second_par1}")
    
    # Parquet strukturu: PAR1 header ... metadata ... [metadata_size 4bytes] PAR1
    # İkinci PAR1-dən sonrakı 4 byte nədir?
    
    # Əslində ikinci PAR1 footer deyil, data içində ola bilər
    # Gəlin faylı düzgün kəsək
    
    # Strategiya: İkinci PAR1-dən 4 byte SONRA metadata size var
    # Sonra PAR1 footer gəlməlidir
    
    # Parquet faylın SONUNDA metadata olur:
    # [data blocks] [metadata] [4-byte metadata size] [PAR1]
    
    # İkinci PAR1-dən sonra nə var?
    print(f"\nAfter second PAR1 (next 50 bytes):")
    print(data[second_par1:second_par1+50].hex())
    print(repr(data[second_par1:second_par1+50]))
    
    # İkinci PAR1 + 4 byte sonrası
    after_second = data[second_par1+4:second_par1+8]
    possible_metadata_size = struct.unpack('<I', after_second)[0]
    print(f"\nPossible metadata size after second PAR1: {possible_metadata_size}")
    
    # Valid Parquet data harada bitir?
    # İkinci PAR1-dən sonra metadata size + PAR1 gəlməlidir
    valid_parquet_end = second_par1 + 4 + 4 + 4  # PAR1 + size + PAR1
    
    # Amma daha asan: second PAR1-i footer kimi götürək
    # və metadata size-ı əvvəldən hesablayaq
    
    # Yeni strategiya: Second PAR1-ə qədər götür və düzgün footer qur
    
    print("\n--- Attempting to reconstruct valid Parquet ---")
    
    # Metadata second PAR1-dən əvvəl olmalıdır
    # Gəlin second PAR1-dən əvvəlki hissəyə baxaq
    
    before_second = data[second_par1-100:second_par1]
    print(f"\n100 bytes before second PAR1:")
    print(before_second.hex())
    
    # Valid parquet: everything up to and including second PAR1 + metadata info
    # Sadə yol: İkinci PAR1 + 8 byte (size + PAR1)
    
    potential_end = second_par1 + 8  # PAR1 + 4 byte size + maybe more
    
    print(f"\nBytes around position {potential_end}:")
    print(data[potential_end-10:potential_end+10].hex())
    
    # Gəlin flag-ın harada başladığını tapaq
    flag_start = data.find(b'FLAG{')
    print(f"\nFLAG starts at position: {flag_start}")
    print(f"FLAG content: {data[flag_start:].decode('utf-8', errors='ignore')}")
    
    # Valid parquet data flag-dan əvvələ qədərdir
    valid_parquet_data = data[:flag_start]
    
    # Amma bu da kifayət deyil, düzgün footer qurmalıyıq
    # İkinci PAR1 footer olmalıdır
    
    # Yeni yanaşma: İkinci PAR1-dən sonra metadata size-ı tap və validasiya et
    for offset in range(0, 100):
        try:
            potential_size_pos = second_par1 + 4 + offset
            if potential_size_pos + 4 < len(data):
                size_bytes = data[potential_size_pos:potential_size_pos+4]
                metadata_size = struct.unpack('<I', size_bytes)[0]
                
                # Metadata file içində olmalıdır
                metadata_start = potential_size_pos - metadata_size
                
                if metadata_start > 0 and metadata_start < second_par1:
                    # Bu potential valid structure
                    # Valid parquet: başdan metadata_start + metadata + size + PAR1
                    valid_end = potential_size_pos + 4 + 4  # size bytes + PAR1
                    
                    if valid_end < flag_start:  # Flag-dan əvvəl olmalı
                        print(f"\n✓ Potential valid structure at offset {offset}:")
                        print(f"  Metadata start: {metadata_start}")
                        print(f"  Metadata size: {metadata_size}")
                        print(f"  Valid Parquet ends at: {valid_end}")
                        
                        # Extract və test et
                        valid_data = data[:valid_end-4] + b'PAR1'  # Ensure PAR1 footer
                        
                        output = file_path.replace('.parquet', f'_extracted_{offset}.parquet')
                        with open(output, 'wb') as f:
                            f.write(valid_data)
                        
                        # Test
                        try:
                            import pandas as pd
                            df = pd.read_parquet(output)
                            print(f"\n✓✓✓ SUCCESS WITH OFFSET {offset}! ✓✓✓")
                            print(f"File: {output}")
                            print(f"Rows: {len(df)}")
                            print(f"Columns: {df.columns.tolist()}")
                            print(f"\nFirst 5 rows:")
                            print(df.head())
                            
                           
                            print(f"\n FLAG FOUND: {data[flag_start:].decode('utf-8', errors='ignore').strip()}")
                            
                            return output, data[flag_start:].decode('utf-8', errors='ignore').strip()
                        except Exception as e:
                            print(f"  ✗ Failed: {str(e)[:100]}")
                            os.remove(output)
        except:
            pass
    
    # Əgər heç nə işləməsə, sadəcə flag-a qədər kəs
    print("\n--- Simple truncation at FLAG position ---")
    output = file_path.replace('.parquet', '_truncated.parquet')
    
    # FLAG-dan əvvəl son PAR1-i tap
    last_par1_before_flag = second_par1
    
    # İkinci PAR1-dən 8 byte sonraya qədər götür
    simple_cut = data[:last_par1_before_flag + 8]
    
    with open(output, 'wb') as f:
        f.write(simple_cut)
    
    try:
        import pandas as pd
        df = pd.read_parquet(output)
        print(f"✓ Simple truncation worked!")
        print(f"File: {output}")
        print(f"Rows: {len(df)}")
        print(f"Flag: {data[flag_start:].decode('utf-8', errors='ignore').strip()}")
        return output, data[flag_start:].decode('utf-8', errors='ignore').strip()
    except Exception as e:
        print(f"Simple truncation failed: {e}")
    
    return None, data[flag_start:].decode('utf-8', errors='ignore').strip()

# İŞLƏT
import struct
file_path = r"./data_lake/raw/archive_batch_seismic_readings.parquet"
result, flag = extract_valid_parquet_and_flag(file_path)

print(f"\n{'='*70}")
print(f" YOUR FLAG: {flag}")
print(f"{'='*70}")
```

```python
import struct
import pandas as pd

def final_repair_parquet(file_path):
    """Son və düzgün təmir"""
    
    with open(file_path, 'rb') as f:
        data = f.read()
    
    print(f"File size: {len(data)} bytes")
    
    # İkinci PAR1 position 40040
    second_par1_pos = 40040
    
    # 100 bytes before second PAR1-də "parquet-cpp-arrow version 22.0.0" görürük
    # Bu metadata-nın sonu deməkdir
    
    # Parquet struktur:
    # [PAR1 header] [data pages] [metadata] [4-byte metadata length] [PAR1 footer]
    
    # İkinci PAR1-dən əvvəlki strukturu yoxlayaq
    # Metadata length 4-byte little-endian, PAR1-dən əvvəl
    
    # Position 40040-da PAR1 var
    # Metadata length ondan 4 byte əvvəl olmalıdır: position 40036
    
    metadata_length_pos = second_par1_pos - 4
    metadata_length_bytes = data[metadata_length_pos:metadata_length_pos+4]
    metadata_length = struct.unpack('<I', metadata_length_bytes)[0]
    
    print(f"Metadata length at position {metadata_length_pos}: {metadata_length}")
    
    # Metadata başlangıcı
    metadata_start = metadata_length_pos - metadata_length
    print(f"Metadata starts at: {metadata_start}")
    
    # Valid Parquet structure:
    # [data from 0 to metadata_start] [metadata] [metadata_length] [PAR1]
    
    valid_end = second_par1_pos + 4  # Include the PAR1 footer
    valid_parquet = data[:valid_end]
    
    print(f"Valid Parquet should be: 0 to {valid_end} = {valid_end} bytes")
    
    # Save
    output_path = file_path.replace('.parquet', '_CLEAN.parquet')
    
    with open(output_path, 'wb') as f:
        f.write(valid_parquet)
    
    print(f"\n Created: {output_path}")
    print(f"Size: {len(valid_parquet)} bytes")
    
    # Verify
    print("\n--- Verification ---")
    with open(output_path, 'rb') as f:
        check_data = f.read()
        print(f"Header: {check_data[:4]}")
        print(f"Footer: {check_data[-4:]}")
    
    # Test reading
    try:
        df = pd.read_parquet(output_path)
        print(f"\n✓✓✓ SUCCESS! ✓✓✓")
        print(f"Rows: {len(df)}")
        print(f"Columns: {df.columns.tolist()}")
        print(f"\nFirst 5 rows:")
        print(df.head())
        print(f"\nData types:")
        print(df.dtypes)
        print(f"\nShape: {df.shape}")
        
        return output_path
        
    except Exception as e:
        print(f"\n Still failed: {e}")
        
        # Alternative: Try cutting earlier
        print("\n--- Trying alternative cut points ---")
        
        for cut_offset in [0, -4, -8, -12, -16, -20]:
            try:
                alt_end = second_par1_pos + 4 + cut_offset
                alt_data = data[:alt_end]
                
                alt_output = file_path.replace('.parquet', f'_CLEAN_v{abs(cut_offset)}.parquet')
                with open(alt_output, 'wb') as f:
                    f.write(alt_data)
                
                df = pd.read_parquet(alt_output)
                print(f"\n✓ SUCCESS with offset {cut_offset}!")
                print(f"File: {alt_output}")
                print(f"Rows: {len(df)}, Columns: {len(df.columns)}")
                print(df.head())
                return alt_output
                
            except Exception as e2:
                print(f"Offset {cut_offset}: {str(e2)[:80]}")
                
    return None


file_path = r"./data_lake/raw/archive_batch_seismic_readings.parquet"
clean_file = final_repair_parquet(file_path)

print("\n" + "="*70)
print(" FLAG: FLAG{DATA_ARCHAEOLOGIST_LVL_99}")
print("="*70)
```

```python
# scripts/recover_archive_batch_from_raw.py
# - archive_batch_seismic_readings.parquet açılmır
# - onu SGX (raw bytes) kimi parse edib
# - düzgün Parquet-a çevirib bərpa etmək

import os
import struct
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

# PATH-lar
RAW_FILE = r"./data_lake/raw/archive_batch_seismic_readings.parquet"
OUT_FILE = r"./data_lake/raw/archive_batch_seismic_readings_recovered.parquet"

records = []

with open(RAW_FILE, "rb") as f:
    # --- SGX GLOBAL HEADER (16 bytes) ---
    header = f.read(16)
    if len(header) != 16:
        raise RuntimeError("Header oxuna bilmədi")

    magic, survey_id, trace_count = struct.unpack("<8sII", header)

    # --- TRACE RECORDS (13 bytes each) ---
    while True:
        chunk = f.read(13)
        if len(chunk) < 13:
            break

        try:
            well_id, depth, amplitude, quality = struct.unpack("<IffB", chunk)
            records.append({
                "well_id": well_id,
                "depth_ft": depth,
                "amplitude": amplitude,
                "quality_flag": quality,
                "survey_id": survey_id,
                "source_file": "archive_batch_seismic_readings.parquet"
            })
        except:
            continue

# DataFrame → Parquet
df = pd.DataFrame(records)

if df.empty:
    raise RuntimeError("Heç bir trace bərpa olunmadı")

table = pa.Table.from_pandas(df)
pq.write_table(table, OUT_FILE)

print("Bərpa edildi:")
print(OUT_FILE)
print("Rows:", len(df))
```
